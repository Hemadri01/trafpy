# University

### Inter-Arrival Distribution

```python
dist = tpg.gen_named_val_dist(dist='weibull',
                              params={'_alpha': 0.9, '_lambda': 6000},
                              min_val=1,
                              max_val=80000,
                             # xlim=[0,80000],
                              round_to_nearest=25,
                              show_fig=True,
                              plot_fig=True,
                              print_data=False)
```

### Flow-Size Distribution

```python
dist = tpg.gen_named_val_dist(dist='lognormal',
                              params={'_mu': 7, '_sigma': 2.5},
                              min_val=1,
                              max_val=2e7,
                              round_to_nearest=25,
                              show_fig=True,
                              plot_fig=True,
                              logscale=True,
                              print_data=False)
```
